```python
import pandas as pd
```

```python
df = pd.read_csv("/content/TeachingRatings.csv")
total = len(df)
```

```python
print("Shape of the data:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
print("\nColumn names:", df.columns.tolist())
```

```python
# Q1
count_gt_45 = (df['eval'] > 4.5).sum()
prob_gt_45 = count_gt_45 / total
print("Q1: P(eval > 4.5) =", round(prob_gt_45, 4))
```

```python
# Q2
count_between = ((df['eval'] > 3.5) & (df['eval'] < 4.2)).sum()
prob_between = count_between / total
print("Q2: P(3.5 < eval < 4.2) =", round(prob_between, 4))
```

```python
# Q3
print("\nQ3 Hypotheses:")
print("H0: μ = 12  (mean is not different from historic mean)")
print("Ha: μ ≠ 12  (mean is different from historic mean)")
```

```python
import math
from scipy.stats import norm

mu = 12
sigma = 5.5
n = 36
x_bar = 10.7

z = (x_bar - mu) / (sigma / math.sqrt(n))

p_value = 2 * norm.cdf(-abs(z))

print("Z-value =", z)
print("P-value =", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```
